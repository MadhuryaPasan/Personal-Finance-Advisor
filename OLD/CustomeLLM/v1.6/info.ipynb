```python
from gguf.gguf_reader import GGUFReader
```

```python
gguf_file_path = "gemma-3-1b-it.Q8_0.gguf"
```

```python
reader = GGUFReader(gguf_file_path)
```

```python
total_parameters = 0
for tensor in reader.tensors:
    total_parameters += tensor.n_elements

total_parameters
```

```python
# Convert to millions for a more readable format
total_parameters_million = total_parameters / 1_000_000

# --- Extract other details from the metadata ---
# The GGUFReader automatically parses key-value metadata.
# Common keys include 'tokenizer.ggml.model_file', 'general.architecture', etc.
metadata = {key: field.parts[field.data[0]] for key, field in reader.fields.items()}

architecture = metadata.get('general.architecture', 'Unknown')
tokenizer_name = metadata.get('tokenizer.ggml.model_file', 'Unknown')
file_type = metadata.get('general.file_type', 'Unknown')
# You can check for other keys like quantization, though it's often in the filename

# --- Print the results ---
print(f"--- Details for GGUF file: {gguf_file_path} ---")
print(f"Architecture: {architecture}")
print(f"Tokenizer: {tokenizer_name}")
print(f"File Type: {file_type}")
print(f"Total Parameters: {total_parameters:,.0f}")
print(f"Total Parameters (Millions): {total_parameters_million:,.2f} M")
```

```python
# --- Check the tensor names and calculate total parameters ---
total_parameters = 0
is_lora_adapter = True # Assume it's a LoRA adapter at first

print("\n--- Inspecting Tensors ---")
for i, tensor in enumerate(reader.tensors):
    # Print the name of the first few tensors to inspect the pattern
    if i < 10:
        print(f"Tensor {i}: {tensor.name}")

    # Check for a common base model tensor name to confirm it's a full model
    if 'model.embed_tokens.weight' in tensor.name or 'model.layers' in tensor.name:
        is_lora_adapter = False
        
    total_parameters += tensor.n_elements

# Convert to millions for a more readable format
total_parameters_million = total_parameters / 1_000_000

print("\n" + "-"*40)
print(f"File Name: {gguf_file_path}")
print(f"Total Tensors Found: {len(reader.tensors)}")

if is_lora_adapter:
    print("\nBased on tensor names, this appears to be a standalone LoRA adapter file.")
    print(f"Parameters in LoRA adapter: {total_parameters:,.0f} ({total_parameters_million:,.2f} M)")
else:
    print("\nBased on tensor names, this appears to be a full, merged GGUF model.")
    print(f"Total Parameters: {total_parameters:,.0f} ({total_parameters_million:,.2f} M)")
```
